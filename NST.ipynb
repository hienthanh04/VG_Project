```python
from pathlib import Path

testA_dir = Path("/content/drive/MyDrive/vangogh_style_transfer/data/arles/testA")
testB_dir = Path("/content/drive/MyDrive/vangogh_style_transfer/data/arles/testB")
exp_dir = Path("/content/drive/MyDrive/vangogh_style_transfer/experiments/nst_arles_v1")
exp_dir.mkdir(parents=True, exist_ok=True)

image_exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".jfif"}

content_files = sorted([p for p in testA_dir.iterdir() if p.suffix.lower() in image_exts])
style_files = sorted([p for p in testB_dir.iterdir() if p.suffix.lower() in image_exts])

selected_contents = content_files[:3]
selected_styles = style_files[:3]

with open(exp_dir / "content_list.txt", "w", encoding="utf-8") as f:
    for p in selected_contents:
        f.write(str(p) + "\n")

with open(exp_dir / "style_list.txt", "w", encoding="utf-8") as f:
    for p in selected_styles:
        f.write(str(p) + "\n")

print("Done")
```
